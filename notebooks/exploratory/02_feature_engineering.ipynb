## 1️⃣ Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

print('✅ Libraries imported!')
```

## 2️⃣ Load Raw Data

```python
df = pd.read_csv('../data/raw/german_credit_original.csv')
df['Risk'] = df['Risk'].map({1: 0, 2: 1})
print(f'✅ Data loaded: {df.shape}')
print(f'Features before engineering: {df.shape[1] - 1}')
```

## 3️⃣ Feature Engineering

```python
df_fe = df.copy()

# 1. Age Groups
df_fe['age_young']  = (df_fe['Age'] < 25).astype(int)
df_fe['age_middle'] = ((df_fe['Age'] >= 25) & (df_fe['Age'] < 60)).astype(int)
df_fe['age_senior'] = (df_fe['Age'] >= 60).astype(int)
print('✅ Age groups created')

# 2. Credit Amount Bins
df_fe['credit_low']    = (df_fe['Credit_Amount'] < 2500).astype(int)
df_fe['credit_medium'] = ((df_fe['Credit_Amount'] >= 2500) & (df_fe['Credit_Amount'] < 5000)).astype(int)
df_fe['credit_high']   = (df_fe['Credit_Amount'] >= 5000).astype(int)
print('✅ Credit amount bins created')

# 3. Duration Bins
df_fe['duration_short']  = (df_fe['Duration'] <= 12).astype(int)
df_fe['duration_medium'] = ((df_fe['Duration'] > 12) & (df_fe['Duration'] <= 24)).astype(int)
df_fe['duration_long']   = (df_fe['Duration'] > 24).astype(int)
print('✅ Duration bins created')

# 4. Financial Ratios
df_fe['credit_duration_ratio']  = df_fe['Credit_Amount'] / (df_fe['Duration'] + 1)
df_fe['credit_age_ratio']       = df_fe['Credit_Amount'] / (df_fe['Age'] + 1)
df_fe['age_credit_interaction'] = df_fe['Age'] * df_fe['Credit_Amount'] / 1000
print('✅ Financial ratios created')

print(f'\nFeatures after engineering: {df_fe.shape[1] - 1}')
```

## 4️⃣ Encoding Categorical Variables

```python
categorical_cols = [
    'Status_Account', 'Credit_History', 'Purpose', 'Savings',
    'Employment', 'Personal_Status', 'Other_Debtors', 'Property',
    'Other_Plans', 'Housing', 'Job', 'Telephone', 'Foreign_Worker'
]

df_encoded = pd.get_dummies(df_fe, columns=categorical_cols,
                            drop_first=False, dtype=int)

print(f'✅ Encoding complete!')
print(f'Final features: {df_encoded.shape[1] - 1}')
print(f'Final shape: {df_encoded.shape}')
```

## 5️⃣ Save Processed Data

```python
df_encoded.to_csv('../data/processed/german_credit_fe_v3.csv', index=False)
print('✅ Processed data saved!')
print(f'Shape: {df_encoded.shape}')
print('\n→ Next: 03_model_training.ipynb')
```
