## 1️⃣ Setup & Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Display settings
pd.set_option('display.max_columns', None)
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('Set2')

print('✅ Libraries imported!')
```

## 2️⃣ Load Data

```python
# Load dataset
column_names = [
    'Status_Account', 'Duration', 'Credit_History', 'Purpose', 'Credit_Amount',
    'Savings', 'Employment', 'Installment_Rate', 'Personal_Status', 'Other_Debtors',
    'Residence_Since', 'Property', 'Age', 'Other_Plans', 'Housing',
    'Existing_Credits', 'Job', 'Num_Dependents', 'Telephone', 'Foreign_Worker', 'Risk'
]

df = pd.read_csv('../data/raw/german_credit_original.csv')
print(f'✅ Dataset loaded: {df.shape}')
df.head()
```

## 3️⃣ Basic Statistics

```python
print('📊 Dataset Info:')
print(f'   Rows: {df.shape[0]}')
print(f'   Columns: {df.shape[1]}')
print(f'   Missing values: {df.isnull().sum().sum()}')
print(f'   Duplicates: {df.duplicated().sum()}')

print('\n📈 Numerical Statistics:')
df.describe()
```

## 4️⃣ Target Variable Analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Distribution
risk_counts = df['Risk'].map({1: 'Bad', 2: 'Good'}).value_counts()
axes[0].pie(risk_counts.values, labels=risk_counts.index,
           colors=['#e74c3c', '#2ecc71'], autopct='%1.1f%%', startangle=90)
axes[0].set_title('Risk Distribution', fontsize=14, fontweight='bold')

# Count plot
risk_counts.plot(kind='bar', ax=axes[1], color=['#e74c3c', '#2ecc71'])
axes[1].set_title('Risk Counts', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Risk Category')
axes[1].set_ylabel('Count')
axes[1].tick_params(rotation=0)

plt.tight_layout()
plt.savefig('../reports/figures/target_distribution.png', dpi=150, bbox_inches='tight')
plt.show()

print(f'Bad Risk: {(df["Risk"]==1).sum()} (70%)')
print(f'Good Risk: {(df["Risk"]==2).sum()} (30%)')
```

## 5️⃣ Numerical Features Analysis

```python
numerical_cols = ['Age', 'Credit_Amount', 'Duration', 'Installment_Rate']

fig, axes = plt.subplots(2, 4, figsize=(20, 10))

for i, col in enumerate(numerical_cols):
    # Histogram
    axes[0, i].hist(df[col], bins=20, color='#006341', alpha=0.7)
    axes[0, i].set_title(f'{col} Distribution')
    axes[0, i].set_xlabel(col)

    # Box plot by risk
    df.boxplot(column=col, by='Risk', ax=axes[1, i])
    axes[1, i].set_title(f'{col} by Risk')

plt.suptitle('Numerical Features Analysis', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('../reports/figures/numerical_features.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 6️⃣ Correlation Analysis

```python
# Correlation matrix
num_df = df[['Age', 'Credit_Amount', 'Duration',
             'Installment_Rate', 'Residence_Since',
             'Existing_Credits', 'Num_Dependents', 'Risk']]

corr_matrix = num_df.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, fmt='.2f',
            cmap='RdYlGn', center=0,
            square=True, linewidths=0.5)
plt.title('Correlation Matrix', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('../reports/figures/correlation_matrix.png', dpi=150, bbox_inches='tight')
plt.show()

print('\n🎯 Correlation with Risk:')
print(corr_matrix['Risk'].sort_values(ascending=False))
```
